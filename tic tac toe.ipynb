```python
import math
board = [" "] * 9
def show():
    print(f"{board[0]} | {board[1]} | {board[2]}")
    print("--+---+--")
    print(f"{board[3]} | {board[4]} | {board[5]}")
    print("--+---+--")
    print(f"{board[6]} | {board[7]} | {board[8]}")
def winner():
    wins = [(0,1,2),(3,4,5),(6,7,8),
            (0,3,6),(1,4,7),(2,5,8),
            (0,4,8),(2,4,6)]
    
    for a,b,c in wins:
        if board[a] == board[b] == board[c] != " ":
            return board[a]
    return "Draw" if " " not in board else None
def minimax(maximizing):
    result = winner()
    if result == "O": return 1
    if result == "X": return -1
    if result == "Draw": return 0
    scores = []
    for i in range(9):
        if board[i] == " ":
            board[i] = "O" if maximizing else "X"
            scores.append(minimax(not maximizing))
            board[i] = " "
    return max(scores) if maximizing else min(scores)
def ai_move():
    best_score = -math.inf
    best_move = 0
    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            score = minimax(False)
            board[i] = " "
            if score > best_score:
                best_score = score
                best_move = i
    board[best_move] = "O"

print("TIC-TAC-TOE AI")
print("You = X | AI = O")
while True:
    show()
    move = int(input("Choose position (1-9): ")) - 1
    if board[move] != " ":
        print("Position already taken!")
        continue
    board[move] = "X"
    if winner():
        break
    ai_move()
    if winner():
        break
show()
result = winner()
if result == "X":
    print("You Win! 🎉")
elif result == "O":
    print("AI Wins! 🤖")
else:
    print("It's a Draw! 🤝")
```
