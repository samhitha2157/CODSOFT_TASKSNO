```python
# Task 4: Movie Recommendation System

movies = {
    "Inception": ["sci-fi", "action", "thriller"],
    "Interstellar": ["sci-fi", "adventure", "drama"],
    "The Dark Knight": ["action", "crime", "thriller"],
    "Titanic": ["romance", "drama"],
    "Avengers": ["action", "sci-fi", "adventure"],
    "The Notebook": ["romance", "drama"],
    "Jurassic Park": ["sci-fi", "adventure", "thriller"],
    "Jumanji": ["adventure", "comedy", "fantasy"]
}
print("🎬 Movie Recommendation System")
print("Available genres: action, sci-fi, thriller, romance, drama, adventure, comedy, fantasy")
genre = input("\nEnter your favourite genre: ").lower()
recommendations = []
for movie, genres in movies.items():
    if genre in genres:
        recommendations.append(movie)
if recommendations:
    print("\n🍿 Recommended Movies:")
    for movie in recommendations:
        print("⭐", movie)
else:
    print("\nSorry, no movies found for that genre.")
```
